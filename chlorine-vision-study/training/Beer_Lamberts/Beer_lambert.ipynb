```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('features_data.csv')
df['absorbance'] = np.log(df['ratio_R_G'])
plt.scatter(df['concentration'], df['absorbance'])
plt.xlabel("Concentration")
plt.ylabel("Absorbance")
plt.show()
```
